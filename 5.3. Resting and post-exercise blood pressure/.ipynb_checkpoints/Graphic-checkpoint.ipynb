```python
import pandas as pd
import matplotlib.pyplot as plt
import os

data_folder = '../5.3. Resting and post-exercise blood pressure'
files = ['rest_Mas.csv', 'rest_Pet.csv', 'rest_Pod.csv', 'fitness_Mas.csv', 'fitness_Pet.csv', 'fitness_Pod.csv']

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

for f in files:
    filepath = os.path.join(data_folder, f)
    
    if not os.path.exists(filepath):
        continue
        
    df = pd.read_csv(filepath)
    t = df.iloc[:, 0].values
    p = df.iloc[:, 1].values

    if 'rest' in f:
        ax1.plot(t, p, label=f, alpha=0.8)
    elif 'fitness' in f:
        ax2.plot(t, p, label=f, alpha=0.8)

ax1.set_title('Давление в манжете, в покое')
ax1.set_xlabel('Время [с]')
ax1.set_ylabel('Давление [мм рт. ст.]')
ax1.set_xlim(0, 60)
ax1.set_ylim(0, 280)
ax1.legend()
ax1.grid(True, linestyle='--', alpha=0.7)

ax2.set_title('Давление в манжете, после нагрузки')
ax2.set_xlabel('Время [с]')
ax2.set_ylabel('Давление [мм рт. ст.]')
ax2.set_xlim(0, 60)
ax2.set_ylim(0, 280)
ax2.legend()
ax2.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.savefig('pressure_plots.png', bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os

# Укажи правильный путь к папке с калибровочными файлами
calib_folder = '../5.2 Blood pressure monitor pressure sensor calibration'

files = ['40_2.csv', '80_1.csv', '120.csv', '160.csv']
pressures = [40, 80, 120, 160]
adc_means = []

for f in files:
    filepath = os.path.join(calib_folder, f)
    
    if not os.path.exists(filepath):
        print(f"Нет файла {filepath}")
        continue
        
    df = pd.read_csv(filepath)
    adc_means.append(df.iloc[:, 1].mean())


if len(adc_means) == 0:
    print("ОШИБКА: Не найдено ни одного файла. Проверь путь calib_folder.")
else:
    adc_means = np.array(adc_means)
    pressures = np.array(pressures)

    
    k, b = np.polyfit(adc_means, pressures, 1)

    plt.figure(figsize=(10, 6))


    plt.scatter(adc_means, pressures, color='red', s=100, zorder=5, label='Измеренные точки')

    x_line = np.linspace(min(adc_means) - 0.02, max(adc_means) + 0.02, 100)
    y_line = k * x_line + b
    plt.plot(x_line, y_line, color='blue', linestyle='--', label='Линия тренда')


    equation_text = f"P = {k:.2f} * N + ({b:.2f})"
    plt.plot([], [], ' ', label=equation_text)

    plt.title('График калибровки датчика давления')
    plt.xlabel('Средний отсчёт АЦП (Напряжение)')
    plt.ylabel('Давление [мм рт. ст.]')
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    

    plt.savefig('pressure-calibration.png', bbox_inches='tight')
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

filepath = '../5.3. Resting and post-exercise blood pressure/rest_Mas.csv'

df = pd.read_csv(filepath)
t = df.iloc[:, 0].values
p = df.iloc[:, 1].values

# Отрезаем начало, где ты качаешь манжету (убираем всё, что выше 180)
mask = p <= 180
t_clean = t[mask]
p_clean = p[mask]

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(t_clean, p_clean, label='давление в манжете', color='#1f77b4')

ax.set_title('Давление в манжете, в покое')
ax.set_xlabel('Время [с]')
ax.set_ylabel('Давление [мм рт. ст.]')

ax.set_xlim(0, 60)
ax.set_ylim(0, 200)

# Включаем мелкую сетку как на образце
ax.minorticks_on()
ax.grid(which='major', color='gray', linestyle='-', linewidth=0.5)
ax.grid(which='minor', color='lightgray', linestyle=':', linewidth=0.5)

ax.legend(loc='upper right')

plt.tight_layout()
plt.savefig('rest_pressure1.png', bbox_inches='tight')
plt.show()
```

```python

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.signal import find_peaks
from scipy.ndimage import median_filter, uniform_filter1d


# ==========================================
# НАСТРОЙКИ
# ==========================================

data_folder = '../5.3. Resting and post-exercise blood pressure'
plots_folder = 'plots'

os.makedirs(plots_folder, exist_ok=True)

files = [
    'rest_Mas.csv',
    'rest_Pet.csv',
    'rest_Pod.csv',
    'fitness_Mas.csv',
    'fitness_Pet.csv',
    'fitness_Pod.csv'
]

# Если начало пульсовых волн известно по графику,
# укажи время первой волны в секундах.
# None = автоматическое определение.

wave_start = {
    'rest_Mas.csv': None,
    'rest_Pet.csv': None,
    'rest_Pod.csv': None,
    'fitness_Mas.csv': None,
    'fitness_Pet.csv': None,
    'fitness_Pod.csv': None
}


# ==========================================
# ОБРАБОТКА ОДНОЙ ЗАПИСИ
# ==========================================

def process_record(filename):

    filepath = os.path.join(data_folder, filename)

    if not os.path.exists(filepath):
        print('Файл не найден:', filepath)
        return

    df = pd.read_csv(filepath)

    time = df.iloc[:, 0].to_numpy(dtype=float)
    p_raw = df.iloc[:, 1].to_numpy(dtype=float)

    # Удаляем некорректные значения
    valid = np.isfinite(time) & np.isfinite(p_raw)
    time = time[valid]
    p_raw = p_raw[valid]

    # Сортировка по времени
    order = np.argsort(time)
    time = time[order]
    p_raw = p_raw[order]

    # Начало времени с нуля
    time = time - time[0]

    if len(time) < 100:
        print(filename, '- недостаточно данных')
        return

    duration = time[-1]
    rate = 1 / np.median(np.diff(time))

    # ======================================
    # 1. СГЛАЖИВАНИЕ ДАВЛЕНИЯ 50 мс
    # ======================================

    window = max(1, int(rate * 0.05))

    p = uniform_filter1d(
        p_raw,
        size=window,
        mode='nearest'
    )

    # Начало спуска давления
    max_idx = np.argmax(p)

    # ======================================
    # 2. ВЫЧИТАНИЕ ТРЕНДА 1.5 СЕКУНДЫ
    # ======================================

    window_trend = max(1, int(rate * 1.5))

    trend = uniform_filter1d(
        p,
        size=window_trend,
        mode='nearest'
    )

    osc = p - trend

    # ======================================
    # 3. ПОИСК ПИКОВ И ВПАДИН
    # ======================================

    distance = max(1, int(0.4 * rate))

    peaks, _ = find_peaks(
        osc[max_idx:],
        distance=distance,
        prominence=0.3
    )

    troughs, _ = find_peaks(
        -osc[max_idx:],
        distance=distance,
        prominence=0.3
    )

    peaks += max_idx
    troughs += max_idx

    # ======================================
    # 4. АМПЛИТУДА ПУЛЬСОВЫХ ВОЛН
    # ======================================

    amp = []
    amp_time = []
    amp_pressure = []

    for a, b in zip(troughs[:-1], troughs[1:]):

        if b <= a + 1:
            continue

        top = a + np.argmax(osc[a:b])

        amplitude = osc[top] - osc[a]

        amp.append(amplitude)
        amp_time.append(time[top])
        amp_pressure.append(p[top])

    amp = np.array(amp)
    amp_time = np.array(amp_time)
    amp_pressure = np.array(amp_pressure)

    if len(amp) < 8:
        print(filename, '- недостаточно пульсовых волн')
        return

    # ======================================
    # 5. МЕДИАННОЕ СГЛАЖИВАНИЕ ПО 7 ВОЛНАМ
    # ======================================

    amp_smooth = median_filter(
        amp,
        size=7,
        mode='nearest'
    )

    max_amp_idx = np.argmax(amp_smooth)
    max_amp = amp_smooth[max_amp_idx]

    # ======================================
    # 6. ОПРЕДЕЛЕНИЕ НАЧАЛА ПУЛЬСОВЫХ ВОЛН
    # ======================================

    manual_start = wave_start[filename]

    if manual_start is not None:

        start_idx = np.searchsorted(
            amp_time,
            manual_start
        )

    else:

        # Автоматический поиск начала устойчивых волн
        threshold = max(
            0.3,
            0.12 * max_amp
        )

        start_idx = 0

        for i in range(len(amp_smooth) - 2):

            if np.all(
                amp_smooth[i:i+3] > threshold
            ):
                start_idx = i
                break

    # ======================================
    # 7. СИСТОЛИЧЕСКОЕ ДАВЛЕНИЕ
    # ======================================

    # Пятая волна после появления пульсаций
    sys_idx = start_idx + 4

    if sys_idx >= len(amp):
        print(filename, '- невозможно определить систолу')
        return

    t_sys = amp_time[sys_idx]
    p_sys = amp_pressure[sys_idx]

    # ======================================
    # 8. ДИАСТОЛИЧЕСКОЕ ДАВЛЕНИЕ
    # ======================================

    dia_idx = None

    candidates = np.where(
        amp_smooth[max_amp_idx + 1:] < 0.8 * max_amp
    )[0]

    if len(candidates) > 0:

        dia_idx = (
            max_amp_idx + 1 + candidates[0]
        )

        t_dia = amp_time[dia_idx]
        p_dia = amp_pressure[dia_idx]

    # ======================================
    # 9. ПУЛЬС НА УЧАСТКЕ 10-30 СЕКУНД
    # ======================================

    inside = peaks[
        (time[peaks] > 10) &
        (time[peaks] < 30)
    ]

    if len(inside) >= 2:

        pulse = (
            (len(inside) - 1) * 60 /
            (time[inside[-1]] - time[inside[0]])
        )

    else:
        pulse = None

    # ======================================
    # 10. ВЫВОД РЕЗУЛЬТАТОВ
    # ======================================

    print('\n' + '-' * 40)
    print(filename)

    if pulse is not None:
        print(f'Пульс: {pulse:.0f} уд/мин')
    else:
        print('Пульс: недостаточно данных')

    print(f'Систолическое давление: {p_sys:.0f} мм рт. ст.')

    if dia_idx is not None:
        print(f'Диастолическое давление: {p_dia:.0f} мм рт. ст.')
        print(f'Давление: {p_sys:.0f}/{p_dia:.0f} мм рт. ст.')
    else:
        print('Диастолическое давление: не определено')

    if manual_start is None:
        print(
            f'Автоматическое начало волн: '
            f'{amp_time[start_idx]:.2f} с'
        )

    if np.max(p) > 200:
        print('ВНИМАНИЕ: давление выше 200 мм рт. ст.!')
        print('Проверь калибровку и исходные данные.')

    # ======================================
    # 11. ГРАФИК ДАВЛЕНИЯ
    # ======================================

    prefix = filename.replace('.csv', '')

    fig, ax = plt.subplots(figsize=(12, 6))

    ax.plot(
        time,
        p,
        color='blue',
        linewidth=1,
        label='Давление'
    )

    ax.plot(
        t_sys,
        p_sys,
        'r*',
        markersize=15,
        label=f'Систола: {p_sys:.0f}'
    )

    ax.annotate(
        'Систола',
        (t_sys, p_sys),
        xytext=(8, 10),
        textcoords='offset points'
    )

    if dia_idx is not None:

        ax.plot(
            t_dia,
            p_dia,
            'g*',
            markersize=15,
            label=f'Диастола: {p_dia:.0f}'
        )

        ax.annotate(
            'Диастола',
            (t_dia, p_dia),
            xytext=(8, 10),
            textcoords='offset points'
        )

    ax.set_title(f'Давление в манжете ({prefix})')
    ax.set_xlabel('Время, с')
    ax.set_ylabel('Давление, мм рт. ст.')

    # Нормальная шкала давления
    ax.set_ylim(
        0,
        max(180, np.ceil(np.max(p) / 20) * 20 + 10)
    )

    ax.set_xlim(0, duration)

    ax.grid(True, linestyle='--', alpha=0.6)
    ax.legend()

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            plots_folder,
            prefix + '-pressure.png'
        ),
        dpi=150
    )

    plt.close(fig)

    # ======================================
    # 12. ГРАФИК ПУЛЬСАЦИЙ
    # ======================================

    fig, ax = plt.subplots(figsize=(12, 6))

    ax.plot(
        time,
        osc,
        color='blue',
        linewidth=0.8,
        label='Пульсовые волны'
    )

    ax.plot(
        time[peaks],
        osc[peaks],
        'r.',
        markersize=5,
        label='Пики'
    )

    ax.set_title(f'Пульсации ({prefix})')
    ax.set_xlabel('Время, с')
    ax.set_ylabel('Пульсации, мм рт. ст.')

    ax.set_xlim(0, duration)

    ax.grid(True, linestyle='--', alpha=0.6)
    ax.legend()

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            plots_folder,
            prefix + '-pulse.png'
        ),
        dpi=150
    )

    plt.close(fig)

    # ======================================
    # 13. ГРАФИК АМПЛИТУДЫ
    # ======================================

    fig, ax = plt.subplots(figsize=(12, 6))

    ax.plot(
        amp_time,
        amp,
        'o',
        color='lightblue',
        label='Амплитуда волн'
    )

    ax.plot(
        amp_time,
        amp_smooth,
        color='blue',
        label='Медиана по 7 волнам'
    )

    ax.axhline(
        0.8 * max_amp,
        color='green',
        linestyle='--',
        label='80% максимума'
    )

    ax.plot(
        amp_time[sys_idx],
        amp_smooth[sys_idx],
        'r*',
        markersize=15,
        label='Систола'
    )

    if dia_idx is not None:

        ax.plot(
            amp_time[dia_idx],
            amp_smooth[dia_idx],
            'g*',
            markersize=15,
            label='Диастола'
        )

    ax.set_title(f'Амплитуда пульсовых волн ({prefix})')
    ax.set_xlabel('Время, с')
    ax.set_ylabel('Амплитуда, мм рт. ст.')

    ax.set_xlim(0, duration)

    ax.grid(True, linestyle='--', alpha=0.6)
    ax.legend()

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            plots_folder,
            prefix + '-amplitude.png'
        ),
        dpi=150
    )

    plt.close(fig)


# ==========================================
# ЗАПУСК ОБРАБОТКИ
# ==========================================

print('\nЗАПИСИ В ПОКОЕ')

for filename in files:
    if filename.startswith('rest'):
        process_record(filename)


print('\nЗАПИСИ ПОСЛЕ НАГРУЗКИ')

for filename in files:
    if filename.startswith('fitness'):
        process_record(filename)


print('\nОбработка завершена!')
print('Графики сохранены в папке:', plots_folder)
```
